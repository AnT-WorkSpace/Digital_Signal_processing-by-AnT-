## 1. What is a Jupyter notebook?

A notebook is a single document that mixes **code**, the **output** that code produces, and **formatted
text** (explanations, headings, formulas) in one place, arranged as a sequence of *cells*. That's the main
difference from the `.py` scripts you've used so far (Programming I), which are plain code with no narrative and no stored output.

Two pieces of vocabulary you'll need constantly:

- A **cell** is one block in the notebook — either a **code cell** (Python code you run) or a **Markdown cell**
  (formatted text you render).
- The **kernel** is the running Python process behind the notebook. It executes your code cells and keeps
  track of every variable and function you've defined — we'll come back to this in detail below, since it's
  the single biggest source of notebook-specific bugs.

## 2. Cells

- A **code cell** runs Python. Run it with **Shift+Enter** (run and move to the next cell) or **Ctrl+Enter**
  (run and stay). The cell's output — printed text, an error, or the value of its last line — appears directly
  underneath it.
- A **Markdown cell** renders formatted text when you run it, instead of executing code.
- The active cell has a coloured outline (typically blue). In command mode the cell has a greyish background.
- If the code has been modified since the last run, the cell will be marked orange.
- To the left of every code cell you'll see a label like `In [ ]`, `In [*]`, or `In [3]`. Empty means it
  hasn't run yet; the asterisk means it's currently running; a number tells you **the order in which cells
  were executed on the kernel** — not their order on the page. Keep an eye on these numbers: if they're out of
  sequence, your notebook was not run top to bottom, and you should not trust its results yet.

## 3. The kernel remembers everything — in execution order, not page order

This is the part that causes real problems every year, so read carefully.

Because the kernel keeps every variable and function you've run in memory, **a notebook does not behave like
a script that always runs top to bottom once**. You can run cells out of order, skip a cell, run one cell
five times, or edit a cell *after* you've already run the ones below it — and the kernel will keep going with
whatever is currently in memory, which may no longer match what's printed on the page.

Try the demo below.

```python
fs = 8000
f0 = 440
period_samples = fs // f0
print('period_samples:', period_samples)
```

Run the cell above. Now imagine that, further down while testing something unrelated,
you quickly reuse the same variable name:

```python
period_samples = 20  # quick manual test — forgot this variable name was already in use
print(period_samples)
```

Run that cell too. Now, **without re-running the first cell above**, add a fresh cell
here and just print `period_samples` again:

```python
print(period_samples)
```

```python
print('period_samples is now', period_samples)
```

The result is `20`, not `18` — even though the first cell on the page still visually
"defines" `period_samples` as `fs // f0`. The kernel doesn't re-derive that value every time you look at the
page; it simply holds whatever the *last executed* cell assigned to that name, regardless of which cell that
was or where it sits on the page.

**This is exactly the kind of bug that can cause headaches**: the notebook can look correct top to bottom while the kernel's actual state depends on the order cells were run in, not the order they're written in.

From the top **Run** and **Kernel menues** you have three commands that matter here:
- **Run** - "play button": run the cell and move to next cell (shift+enter)
- **Restart** — "a circular arrow icon": wipes all variables and functions, kernel starts fresh (empty memory).
- **Restart & Run All** — "Fast Forward": removes all displayed outputs from the page and then re-runs every cell top to bottom in order.

In case your results are not plausible or weird, **Restart & Run All** and confirm every cell still reproduces the same output from a clean kernel.

### Multiple plots in a single figure
It is often more reasonable to have several plots within the same figure 
(this also improves the look of your final thesis!). 
You can do this using a (sadly) slightly different apprach in python. 

When you have more than one axes, you work with an **array of axes objects** instead of 
calling `plt` functions directly. The commands change slightly but sadly noticably:

| Single plot         | Subplots (per axes)      |
|----------------------|---------------------------|
| `plt.plot()`         | `ax.plot()`               |
| `plt.xlabel()`        | `ax.set_xlabel()`         |
| `plt.ylabel()`        | `ax.set_ylabel()`         |
| `plt.title()`         | `ax.set_title()`          |
| `plt.grid()`          | `ax.grid()`               |
| `plt.xlim()`          | `ax.set_xlim()`           |
| `plt.ylim()`          | `ax.set_ylim()`           |
| `plt.xticks()`        | `ax.set_xticks()` / `ax.set_xticklabels()` |

You can add a title to the collection of plots using 
`fig.suptitle("Title of Plot", fontsize=14)`
`fig.subplots_adjust(top=0.85)  # make room for suptitle`

```python
fig, axs = plt.subplots(3, 1, figsize=(6, 7))   # 3 vertical subplots 

ax=axs[0]
ax.stem(n, impulse, basefmt="gray")
ax.set_xlim(-5.5, 15.5)
ax.set_ylim(-0.2, 1.2)
ax.set_xticks(n)
ax.set_xlabel("$n$")
ax.set_ylabel("$\\delta[n]$")
ax.grid()

ax=axs[1]
ax.stem(n, step, basefmt="gray")
ax.set_xlim(-5.5, 15.5)
ax.set_ylim(-0.2, 1.2)
ax.set_xticks(n)
ax.set_xlabel("$n$")
ax.set_ylabel("$u[n]$")
ax.grid()

ax=axs[2]
ax.stem(n, ramp, basefmt="gray")
ax.set_xlim(-5.5, 15.5)
ax.set_ylim(-0.2, 15.5)
ax.set_xticks(n)
ax.set_xlabel("$n$")
ax.set_ylabel("$r[n]$")
ax.grid()

fig.suptitle("3 Plots in a Single Figure", fontsize=14)
fig.subplots_adjust(top=0.85)   # make room for suptitle
plt.tight_layout()
plt.show()
```

### Why the difference?

`plt.plot()` etc. implicitly act on "whatever the current axes is." That works fine with 
one plot, but with several axes you need to tell Matplotlib *which* one you mean — hence 
`ax.plot()`, `ax.set_xlabel()`, and so on, called explicitly on each `Axes` object (`axs[0]`, `axs[1]`, ...).
